```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Diwali Sales Data.csv",encoding="unicode_escape")
df
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
#drop blank columns
df.drop(["Status","unnamed1"],axis=1,inplace=True)
```

```python
pd.isnull(df)
```

```python
#check for null values
pd.isnull(df).sum()
```

```python
df['Amount']=df["Amount"].fillna(df["Amount"].median())
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
# change data type
df["Amount"]=df["Amount"].astype("int")
```

```python
df["Amount"].dtypes
```

```python
df.columns
```

```python
df.rename(columns={"Marital_Status":"Married"})
```

```python
df.describe()
```

# Exploratory Data Analysis

```python
#Gender
sns.countplot(x="Gender",data=df,color="teal")
sns.set(rc={"figure.figsize":(10,5)})
```

```python
df.groupby(["Gender"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
```

```python
df1=df.groupby(["Gender"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(5,5)})
sns.barplot(x="Gender",y="Amount",data=df1,hue="Gender")
plt.title("Gender Wise Amount")
```

From Above graph we can see that most of the buyers are females and even the purchasing power of females are greater than males.

# Age Group 

```python
df1=sns.countplot(x="Age Group",data=df,hue="Gender")
sns.set(rc={"figure.figsize":(10,5)})
plt.title("Age Group")
for bars in df1.containers:
    df1.bar_label(bars)
```

```python
df.groupby("Age Group")["Orders"].sum().plot(kind="bar",color=["r","teal","purple","gold","yellow","green","k"])
sns.set(rc={"figure.figsize":(10,5)})
plt.title("Age Group Wise Order")
```

From Above graph it is observed that the 26-35 age group has the highest number of orders in the Diwali sales.

```python
df1=df.groupby(["Age Group"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(10,5)})
sns.barplot(x="Age Group",y="Amount",data=df1,hue="Age Group")
plt.title("Age Group wise Amount")
```

 For above graph we can see the buyers are of age group between 26-35 years female

# State Wise Orders

```python
df1=df.groupby(["State"],as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False)
sns.set(rc={"figure.figsize":(25,5)})
plt.title("State Wise Orders")
sns.barplot(x="State",y="Orders",data=df1,hue="State")
```

```python
df1=df.groupby(["State"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(25,5)})
plt.title("State Wise Amount")
sns.barplot(x="State",y="Amount",data=df1,hue="State")
```

From above graph we can see that most of the orders & Total sales Amount are from Uttar pradesh,Maharashtra & Karnataka

# Zones Wise Orders

```python
sns.countplot(x="Zone",data=df,hue="Zone")
sns.set(rc={"figure.figsize":(5,5)})
plt.title("Zone Wise Orders")
plt.show()
```

```python
df1=df.groupby(["Zone"],as_index=False)["Marital_Status"].sum().sort_values(by="Marital_Status",ascending=False)
sns.set(rc={"figure.figsize":(5,5)})
plt.title("Zone Wise Marital_Status")
sns.barplot(x="Zone",y="Marital_Status",data=df1,hue="Zone")
```

```python
df1=df.groupby(["Zone"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(5,5)})
sns.barplot(x="Zone",y="Amount",data=df1,hue="Zone")
plt.title("Zone wise Amount")
```

```python
df1=df.groupby(["Zone","Gender"],as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False)
sns.set(rc={"figure.figsize":(5,5)})
sns.barplot(x="Zone",y="Orders",data=df1,hue="Gender")
```

From Above Graph we can see that most of the orders are from Central Zone

# Marital Status

```python
sns.countplot(x="Marital_Status",data=df,hue="Marital_Status")
plt.show()
```

```python
df1=df.groupby(["Marital_Status","Gender"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(5,5)})
sns.barplot(x="Marital_Status",y="Amount",data=df1,hue="Gender")
plt.title("Marital Status vs Amount")
```

From Above Graph we can see that most of the buyers are married Women & they have high purchasing power*

# Occupation

```python
df1=sns.countplot(x="Occupation",data=df,hue="Occupation")
sns.set(rc={"figure.figsize":(25,5)})
for bars in df1.containers:
    df1.bar_label(bars)
plt.title("Occupation vs Count")
```

```python
df1=df.groupby(["Occupation"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(25,5)})
sns.barplot(x="Occupation",y="Amount",data=df1,hue="Occupation")
plt.title("Occupation vs Amount")
```

From Above Graph we can see that most of the buyers are working in IT sector,HealthCare & Aviation sector

# Product Category

```python
#Product Category
df1=sns.countplot(x="Product_Category",data=df,hue="Product_Category")
sns.set(rc={"figure.figsize":(25,5)})
for bars in df1.containers:
    df1.bar_label(bars)
plt.title("Product Category vs Count")
```

```python
df1=df.groupby(["Product_Category"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.set(rc={"figure.figsize":(35,10)})
sns.barplot(x="Product_Category",y="Amount",data=df1,hue="Product_Category")
plt.title("Product Category vs Amount")
```

From above graph we can see that most of the sold products are from Food,Clothing&Apparel,Electronics & Gadgets

# Conclusion:

The Diwali Sales Analysis project provides clear insights into customer behavior and sales patterns during the festive season. The analysis reveals that the 26–35 age group is the most active, contributing the highest number of orders. It is also observed that female customers have higher purchasing participation compared to males.

Additionally, certain states and product categories contribute significantly to overall revenue, showing that location and product preference play a crucial role in sales performance. Marital status and occupation also influence buying behavior, indicating that customers with stable income tend to spend more.
